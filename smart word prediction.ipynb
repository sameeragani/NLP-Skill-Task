```python
import re
from collections import defaultdict, Counter

# -------------------------------
# Training Text
# -------------------------------

text = """
Artificial intelligence is changing the world.
Artificial intelligence helps computers learn from data.
Machine learning is a part of artificial intelligence.
Machine learning allows computers to learn automatically.
Deep learning is a type of machine learning.
Natural language processing helps computers understand human language.
Natural language processing is used in chatbots and translation.
Artificial intelligence is used in healthcare.
Artificial intelligence is used in education.
Machine learning is used in healthcare and education.
"""

# -------------------------------
# Text Preprocessing
# -------------------------------

text = text.lower()

# Remove punctuation
text = re.sub(r"[^a-zA-Z\s]", "", text)

# Tokenization
words = text.split()

print("Total words:", len(words))


# -------------------------------
# Build Trigram Model
# -------------------------------

model = defaultdict(Counter)

for i in range(len(words) - 2):

    word1 = words[i]
    word2 = words[i + 1]
    next_word = words[i + 2]

    model[(word1, word2)][next_word] += 1


# -------------------------------
# Prediction Function
# -------------------------------

def predict_next_words(sentence, number=3):

    sentence = sentence.lower()

    sentence = re.sub(r"[^a-zA-Z\s]", "", sentence)

    input_words = sentence.split()

    if len(input_words) < 2:
        print("Please enter at least two words.")
        return

    last_two_words = (
        input_words[-2],
        input_words[-1]
    )

    predictions = model.get(last_two_words)

    if not predictions:
        print("\nNo prediction found.")
        print("Try another sentence.")
        return

    print("\nNext Word Predictions:")

    for word, count in predictions.most_common(number):
        print(f"{word}  →  {count}")


# -------------------------------
# User Input
# -------------------------------

while True:

    sentence = input("\nEnter a sentence: ")

    if sentence.lower() == "exit":
        print("Program ended.")
        break

    predict_next_words(sentence)
```
