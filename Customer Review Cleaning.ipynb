```python
# ============================================================
# CUSTOMER REVIEW CLEANING SYSTEM
# NLP TASK - IMDb 50K Movie Reviews
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import re
import string
import pandas as pd
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer


# ------------------------------------------------------------
# 2. DOWNLOAD REQUIRED NLTK DATA
# ------------------------------------------------------------

nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')


# ------------------------------------------------------------
# 3. LOAD THE IMDb 50K DATASET
# ------------------------------------------------------------

# Make sure "IMDB Dataset.csv" is in the same folder
# as this Python file.

df = pd.read_csv("IMDB_Dataset_CLEANED.csv")


# ------------------------------------------------------------
# 4. DISPLAY BASIC DATASET INFORMATION
# ------------------------------------------------------------

print("========================================")
print("IMDb 50K DATASET")
print("========================================")

print("\nFirst 5 rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)


# ------------------------------------------------------------
# 5. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\n========================================")
print("MISSING VALUES")
print("========================================")

print(df.isnull().sum())


# ------------------------------------------------------------
# 6. CHECK DUPLICATE REVIEWS
# ------------------------------------------------------------

print("\n========================================")
print("DUPLICATE REVIEWS")
print("========================================")

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)


# ------------------------------------------------------------
# 7. REMOVE MISSING VALUES
# ------------------------------------------------------------

df = df.dropna()


# ------------------------------------------------------------
# 8. REMOVE DUPLICATE REVIEWS
# ------------------------------------------------------------

df = df.drop_duplicates()


# ------------------------------------------------------------
# 9. DISPLAY DATASET INFORMATION AFTER CLEANING
# ------------------------------------------------------------

print("\n========================================")
print("DATASET AFTER REMOVING MISSING VALUES")
print("AND DUPLICATES")
print("========================================")

print("Number of rows:", len(df))


# ------------------------------------------------------------
# 10. CREATE STOPWORDS SET
# ------------------------------------------------------------

stop_words = set(stopwords.words('english'))


# ------------------------------------------------------------
# 11. CREATE STEMMER
# ------------------------------------------------------------

stemmer = PorterStemmer()


# ------------------------------------------------------------
# 12. CREATE TEXT CLEANING FUNCTION
# ------------------------------------------------------------

def clean_text(text):

    # ----------------------------------------
    # Step 1: Convert text to lowercase
    # ----------------------------------------

    text = text.lower()


    # ----------------------------------------
    # Step 2: Remove HTML tags
    # Example: <br /> <p> </p>
    # ----------------------------------------

    text = re.sub(r'<.*?>', '', text)


    # ----------------------------------------
    # Step 3: Remove URLs
    # ----------------------------------------

    text = re.sub(r'http\S+|www\S+', '', text)


    # ----------------------------------------
    # Step 4: Remove numbers
    # ----------------------------------------

    text = re.sub(r'\d+', '', text)


    # ----------------------------------------
    # Step 5: Remove punctuation
    # ----------------------------------------

    text = text.translate(
        str.maketrans('', '', string.punctuation)
    )


    # ----------------------------------------
    # Step 6: Remove special characters
    # Keep only letters and spaces
    # ----------------------------------------

    text = re.sub(r'[^a-zA-Z\s]', '', text)


    # ----------------------------------------
    # Step 7: Tokenization
    # Split sentence into words
    # ----------------------------------------

    words = word_tokenize(text)


    # ----------------------------------------
    # Step 8: Remove stopwords
    # ----------------------------------------

    words = [
        word for word in words
        if word not in stop_words
    ]


    # ----------------------------------------
    # Step 9: Stemming
    # ----------------------------------------

    words = [
        stemmer.stem(word)
        for word in words
    ]


    # ----------------------------------------
    # Step 10: Join words back into a sentence
    # ----------------------------------------

    cleaned_text = ' '.join(words)


    # Return cleaned text
    return cleaned_text


# ------------------------------------------------------------
# 13. APPLY CLEANING FUNCTION TO REVIEWS
# ------------------------------------------------------------

df['cleaned_review'] = df['review'].apply(clean_text)


# ------------------------------------------------------------
# 14. DISPLAY ORIGINAL AND CLEANED REVIEWS
# ------------------------------------------------------------

print("\n========================================")
print("ORIGINAL VS CLEANED REVIEWS")
print("========================================")

for i in range(5):

    print("\nReview", i + 1)

    print("Original:")
    print(df['review'].iloc[i])

    print("\nCleaned:")
    print(df['cleaned_review'].iloc[i])

    print("----------------------------------------")


# ------------------------------------------------------------
# 15. DISPLAY FINAL DATASET
# ------------------------------------------------------------

print("\n========================================")
print("FINAL DATASET")
print("========================================")

print(df[['review', 'sentiment', 'cleaned_review']].head())


# ------------------------------------------------------------
# 16. SAVE PREPROCESSED DATASET
# ------------------------------------------------------------

df.to_csv(
    "IMDB_Preprocessed.csv",
    index=False
)


# ------------------------------------------------------------
# 17. FINAL MESSAGE
# ------------------------------------------------------------

print("\n========================================")
print("TASK COMPLETED SUCCESSFULLY")
print("========================================")

print("Preprocessed dataset saved as:")
print("IMDB_Preprocessed.csv")

print("\nFinal columns:")
print(df.columns)
```
