```python
# ============================================================
# SEARCH QUERY SPELLING CORRECTOR
# NLP TASK
# BIRKBECK SPELLING ERROR CORPUS
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT REQUIRED MODULES
# ------------------------------------------------------------

import re
import pandas as pd
import numpy as np


# ------------------------------------------------------------
# 2. LOAD BIRKBECK SPELLING ERROR CORPUS
# ------------------------------------------------------------

file_path = "birkbeck.dat"

with open(file_path, "r", encoding="utf-8") as file:
    lines = file.readlines()


print("========================================")
print("BIRKBECK SPELLING ERROR CORPUS")
print("========================================")

print("Dataset loaded successfully!")
print("Number of lines:", len(lines))


# ------------------------------------------------------------
# 3. BUILD VOCABULARY OF CORRECTLY SPELLED WORDS
# ------------------------------------------------------------

vocabulary = set()

for line in lines:

    line = line.strip()

    # Correct words start with $
    if line.startswith("$"):

        correct_word = line[1:].lower()

        if correct_word:
            vocabulary.add(correct_word)


print("\nVocabulary created successfully!")
print("Number of correct words:", len(vocabulary))


# ------------------------------------------------------------
# 4. CREATE EDIT DISTANCE FUNCTION
# ------------------------------------------------------------

def edit_distance(word1, word2):

    # Create a matrix
    rows = len(word1) + 1
    cols = len(word2) + 1

    distance = np.zeros((rows, cols), dtype=int)


    # Initialize first column
    for i in range(rows):
        distance[i][0] = i


    # Initialize first row
    for j in range(cols):
        distance[0][j] = j


    # Calculate edit distance
    for i in range(1, rows):

        for j in range(1, cols):

            if word1[i - 1] == word2[j - 1]:
                cost = 0
            else:
                cost = 1


            # Insertion
            insertion = distance[i][j - 1] + 1

            # Deletion
            deletion = distance[i - 1][j] + 1

            # Substitution
            substitution = distance[i - 1][j - 1] + cost


            distance[i][j] = min(
                insertion,
                deletion,
                substitution
            )


    return distance[rows - 1][cols - 1]


# ------------------------------------------------------------
# 5. FIND CLOSEST CORRECT WORD
# ------------------------------------------------------------

def find_correction(word):

    best_word = word
    best_distance = float("inf")


    for candidate in vocabulary:

        # Ignore words with very different lengths
        if abs(len(word) - len(candidate)) > 3:
            continue


        distance = edit_distance(word, candidate)


        if distance < best_distance:

            best_distance = distance
            best_word = candidate


    return best_word, best_distance


# ------------------------------------------------------------
# 6. ACCEPT USER SEARCH QUERY
# ------------------------------------------------------------

print("\n========================================")
print("SEARCH QUERY SPELLING CORRECTOR")
print("========================================")

query = input("\nEnter your search query: ")


# ------------------------------------------------------------
# 7. TOKENIZE THE QUERY
# ------------------------------------------------------------

words = re.findall(
    r"[a-zA-Z]+",
    query.lower()
)


# ------------------------------------------------------------
# 8. IDENTIFY INCORRECT WORDS
# ------------------------------------------------------------

incorrect_words = []

for word in words:

    if word not in vocabulary:

        incorrect_words.append(word)


# ------------------------------------------------------------
# 9. FIND CORRECTIONS
# ------------------------------------------------------------

corrections = {}

for word in incorrect_words:

    corrected_word, distance = find_correction(word)

    corrections[word] = corrected_word


# ------------------------------------------------------------
# 10. CREATE CORRECTED QUERY
# ------------------------------------------------------------

corrected_words = []

for word in words:

    if word in corrections:

        corrected_words.append(
            corrections[word]
        )

    else:

        corrected_words.append(word)


corrected_query = " ".join(corrected_words)


# ------------------------------------------------------------
# 11. DISPLAY ORIGINAL QUERY
# ------------------------------------------------------------

print("\n========================================")
print("RESULT")
print("========================================")

print("\nOriginal Query:")
print(query)


# ------------------------------------------------------------
# 12. DISPLAY INCORRECT WORDS
# ------------------------------------------------------------

print("\nIncorrect Words:")

if len(incorrect_words) == 0:

    print("No spelling errors found.")

else:

    for word in incorrect_words:

        print(word)


# ------------------------------------------------------------
# 13. DISPLAY SUGGESTED CORRECTIONS
# ------------------------------------------------------------

print("\nSuggested Corrections:")

if len(corrections) == 0:

    print("No corrections required.")

else:

    for wrong_word, correct_word in corrections.items():

        print(
            wrong_word,
            "→",
            correct_word
        )


# ------------------------------------------------------------
# 14. DISPLAY FINAL CORRECTED QUERY
# ------------------------------------------------------------

print("\nCorrected Query:")
print(corrected_query)


# ------------------------------------------------------------
# 15. FINAL MESSAGE
# ------------------------------------------------------------

print("\n========================================")
print("SPELLING CORRECTION COMPLETED")
print("========================================")
```
